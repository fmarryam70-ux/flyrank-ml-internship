## 2. My model under an honest split (before/after)

*Re-run your Week-5 model under a grouped or time-aware split. Show both numbers.*





My Week 5 model was trained and evaluated using a grouped client split. This validation strategy ensures that pages from the same client do not appear in both the training and testing datasets, reducing information leakage and providing a more realistic estimate of performance on unseen clients.

Since the Week 5 model already used grouped validation, I re-ran the same evaluation to confirm the results under the same honest validation strategy.

```python
import pandas as pd

comparison = pd.DataFrame({
    "Model": [
        "Logistic Regression",
        "Decision Tree",
        "Random Forest",
        "Base Rate (Random)",
        "Week 4 Rule Baseline"
    ],
    "Precision@20": [
        0.850,
        0.450,
        0.550,
        0.511,
        0.400
    ],
    "Precision@50": [
        0.740,
        0.580,
        0.580,
        0.511,
        0.420
    ]
})

comparison
```

| Decision Tree | 0.450 | 0.580 |

| Week 4 Rule Baseline | 0.400 | 0.420 |

Logistic Regression achieved the highest Precision@20 (0.850) and Precision@50 (0.740), outperforming both the Week 4 rule-based baseline and the other machine learning models.

Using grouped client validation reduces optimistic performance estimates because data from the same client is never shared between the training and testing datasets. This provides a more trustworthy estimate of how the model is expected to perform on previously unseen clients.

Because the Week 5 notebook already used grouped validation, re-running the model confirmed the same evaluation approach rather than introducing a new validation strategy.

## 3. Leakage audit

*The same hunt from Week 3, on your final feature set.*



The final feature set was reviewed to identify variables that could reveal the target label or provide information unavailable at prediction time.

```python
import pandas as pd

leakage_audit = pd.DataFrame({
    "Feature":[
        "client_id",
        "trend_direction",
        "historical_clicks",
        "historical_impressions",
        "content_age"
    ],
    "Leakage":[
        "Yes",
        "Yes",
        "No",
        "No",
        "No"
    ],
    "Reason":[
        "Identifier only; should not be used as a feature.",
        "Directly used to construct the target label.",
        "Available before prediction.",
        "Available before prediction.",
        "Available before prediction."
    ]
})

leakage_audit
```

### Leakage Summary

The audit identified two variables that could cause leakage.

- **client_id** is only an identifier and should never be used as a predictive feature.
- **trend_direction** directly defines the target label and therefore must not be included as an input feature.

The remaining historical features represent information available before prediction and therefore do not introduce target leakage.

## 4. Claim rewrite

*Take your own boldest sentence and rewrite it in safe language: observed, measured, directional, decision-support.*



The original conclusion from Week 5 was intentionally rewritten using careful, evidence-based language.

```python
original_claim = """
The model accurately predicts declining pages.
"""

revised_claim = """
In this dataset, Logistic Regression achieved the highest Precision@20 (0.850) and Precision@50 (0.740) under grouped client validation. Compared with the Week 4 baseline, the model demonstrated improved ranking performance. These results are observational and the model should be used as a decision-support tool rather than an automated decision-making system.
"""

print("Original Claim:\n")
print(original_claim)

print("\nRevised Claim:\n")
print(revised_claim)
```

### Conclusion

The revised claim avoids overstating model capability. It reports only the measured performance observed on the evaluation dataset and clearly states that the model is intended to support human decision-making rather than replace expert judgment.
