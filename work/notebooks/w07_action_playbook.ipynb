## Cost / Value Considerations

Refreshing every page is not cost-effective. The ranked action queue helps prioritize pages that are expected to provide the greatest value based on the model's predictions.

- High-ranked pages should be reviewed first because they have the highest estimated business impact.
- Medium-ranked pages should be reviewed if resources are available.
- Low-ranked pages should be monitored before investing additional effort.

This prioritization supports efficient use of editorial time while keeping a human reviewer responsible for final decisions.
